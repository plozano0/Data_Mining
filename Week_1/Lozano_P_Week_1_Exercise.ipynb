# Find the top five games by critic score.

```python
# Top five games by critic score
df.nlargest(5, 'Critic_Score')
# Alternative method
# df.sort_values(by='Critic_Score', ascending=False).head(5)
```

> Whoa! Looks like there are duplicate entries for the same game (i.e., Grand Theft Auto IV). Let's do this again...

```python
# Drop duplicates based on the 'Name' column to get unique games
df_unique = df.drop_duplicates(subset='Name')

# Reference the new DataFrame with unique games to pass to the nlargest function
df_unique.nlargest(5, 'Critic_Score')
```

> Much better. Now we can see that Take-Two Interactive's games are doing a phenomenal job in terms of critic scores.
